Project 2: Currency Exchange Rate Tracker
Objective: Build a Python application that retrieves currency exchange rates using an API and stores the data in an SQLite database.

```python
import requests
import sqlite3
import time


fromCur = "gbp"
toCur = "usd"

url = f"https://cdn.jsdelivr.net/npm/@fawazahmed0/currency-api@latest/v1/currencies/{fromCur}.json"

response = requests.get(url)
data = response.json()

toCurVal = data[fromCur][toCur]


def create_database(db_name):
    conn = sqlite3.connect(db_name)
    cursor = conn.cursor()

    # Create 'rates' table for storing the rates
    # Each table design will be unique based on the API you visit
    cursor.execute('''
        CREATE TABLE IF NOT EXISTS rates (
            id INTEGER PRIMARY KEY AUTOINCREMENT,
            fromCur TEXT,
            toCur TEXT,
            rate INTEGER,
            time TEXT
        )
    ''')

    conn.commit()
    return conn, cursor

def insert_rate_to_db(cursor, fromCur, toCur, toCurVal, conn):
    # SQL command to insert data into the rates table
    cursor.execute('''
        INSERT INTO rates (fromCur, toCur, rate, time)
        VALUES (?, ?, ?, ?)
    ''', (
        fromCur,
        toCur,
        toCurVal,
        time.time()
    ))
    conn.commit()
    print("Fact inserted successfully")


conn, cursor = create_database('rates.db')
insert_rate_to_db(cursor, fromCur, toCur, toCurVal, conn)
```

```python
def read_rates_from_db(cursor):
    cursor.execute('SELECT * FROM rates')
    rates = cursor.fetchall()
    return rates

# Example usage

# Connect to the existing database
conn = sqlite3.connect("rates.db")
cursor = conn.cursor()

# Read facts from db
rates = read_rates_from_db(cursor)

# print facts
for rate in rates:
   print(rate)
```
