Question 1: Temperature Analysis
You are provided with a list of daily temperatures (in °C) for a month: temperatures = [18, 22, 19, 25, 17, 21, 23, 20, 26, 24, 19, 22, 27, 21, 23, 18, 24, 25, 19, 20, 22, 26, 23, 21, 19, 24, 22, 20, 18, 23]
Write a Python program to:
a. Calculate and print the number of days with temperatures above 24°C.
b. Identify and print the highest and lowest temperatures of the month.

```python
temperatures = [18, 22, 19, 25, 17, 21, 23, 20, 26, 24, 19, 22, 27, 21, 23, 18, 24, 25, 19, 20, 22, 26, 23, 21, 19, 24, 22, 20, 18, 23]
# Calculate and print the number of days with temperatures above 24°C.
hot_days = 0
for temp in temperatures:
  if temp > 24:
    hot_days += 1
print(f"Number of hot days (above 24°C): {hot_days}")

#Identify and print the highest and lowest temperatures of the month.
highest_temp =  max(temperatures)
lowest_temp = min(temperatures)
print(f"Highest temperature of the month: {highest_temp}°C")
print(f"Lowest temperature of the month: {lowest_temp}°C")

```

Question 2: Student Grades Processing
You have a list of student scores out of 100:

scores = [88, 76, 90, 85, 92, 67, 73, 81, 95, 78]

```python
scores = [88, 76, 90, 85, 92, 67, 73, 81, 95, 78]

# Calculates and prints the average score.
average_score = sum(scores) / len(scores)
print(f"Average score: {average_score}")

# Determines the number of students who passed and failed. (Assume a passing score is 70 or above.)
passed_students = sum(1 for score in scores if score >= 70)
failed_students = sum(1 for score in scores if score < 70)
print(f"Number of students who passed: {passed_students}")
print(f"Number of students who failed: {failed_students}")

# Prints a message for each student indicating "Pass" or "Fail".
for i, score in enumerate(scores, start=1):
  status = "Pass" if score >= 70 else "Fail"
  print(f"Student {i}: {status}")

```

Question 5: Text Processing

You have a list of words:

words = ['python', 'data', 'science', 'machine', 'learning', 'artificial', 'intelligence']
Tasks:

a. Create a new list containing the lengths of each word.
b. Find and print the longest word in the list.
c. Create a list of words that start with the letter 'a'.

```python
words = ['python', 'data', 'science', 'machine', 'learning', 'artificial', 'intelligence']

word_lengths = [len(word) for word in words]

longest_word = max(words, key=len)

a_words = [word for word in words if word.startswith("a")]

print(f"Lengths of words: {word_lengths}")
print(f"Longest word: {longest_word}")
print(f"Words starting with 'a': {a_words}")
```

Question 6: Matrix Addition
Consider the following two 3x3 matrices represented as lists of lists:

matrix_a = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

matrix_b = [
    [9, 8, 7],
    [6, 5, 4],
    [3, 2, 1]
]
Tasks:

Write a Python program to:
a. Add the two matrices and store the result in a new matrix.
b. Print the resulting matrix.

```python
matrix_a = [ [1, 2, 3], [4, 5, 6], [7, 8, 9] ]

matrix_b = [ [9, 8, 7], [6, 5, 4], [3, 2, 1] ]

resulting_matrix = []
for i in range(len(matrix_a)):
  new_row = []
  for j in range(len(matrix_a[i])):
    new_row.append(matrix_a[i][j] + matrix_b[i][j])
  resulting_matrix.append(new_row)

print('Resulting Matrix:')
for row in resulting_matrix:
  print(row)
```
